# ComfyUI instance on Colab

Starts ComfyUI on a Colab GPU and lets you connect from the **Remote Connection** of [Comfy Desktop](https://github.com/Comfy-Org/Comfy-Desktop) or from a regular browser. Models and the environment are kept in `ComfyUI-Colab/` on Google Drive.

**How to use:** Choose a GPU in the runtime type, then run cells **1 → 2 → 3 → one of 6-1 to 6-3** in order. Set the fields and checkboxes of a cell before you run it. An action with a checkbox only happens when you run the cell with the box checked.

**Before you start:** Connecting Drive grants access to your whole Drive. Use a dedicated Google account that holds no personal data, and install only custom nodes and models you trust (see "Security" at the end). Follow the [Colab terms](https://research.google.com/colaboratory/faq.html); a paid plan is assumed.


```python
#@title 0. Check for updates
#@markdown Checks the latest release of ComfyUI. Turn this on to use it instead of `COMFY_REF` in cell 1. 
USE_LATEST_COMFYUI = False #@param {type:"boolean"}
import re, subprocess
found = subprocess.run(["git", "ls-remote", "--tags", "--refs", "https://github.com/Comfy-Org/ComfyUI.git", "v*"],
                       capture_output=True, text=True, timeout=60)
releases = [tuple(map(int, tag.groups())) for tag in
            (re.search(r"refs/tags/v(\d+)\.(\d+)\.(\d+)$", line) for line in found.stdout.splitlines()) if tag]
if found.returncode or not releases:
    raise RuntimeError("Could not get the latest release. Try again in a while.")
LATEST_COMFY_REF = "v" + ".".join(map(str, max(releases)))
print("Latest ComfyUI release:", LATEST_COMFY_REF)
# Cell 1 sets COMFY_REF from its form, so it takes the choice made here from this name.
COMFY_REF_OVERRIDE = LATEST_COMFY_REF if USE_LATEST_COMFYUI else ""
if USE_LATEST_COMFYUI:
    COMFY_REF = LATEST_COMFY_REF
    print("COMFY_REF is set to the latest release. Run cell 1 next.")
else:
    print("COMFY_REF keeps the value entered in cell 1. To use the latest release, check USE_LATEST_COMFYUI and run the cell.")
```

## Security

- Cell 6-1 exposes nothing to the outside. With cells 6-2 and 6-3 only an authenticated entrance is exposed: 6-2 rejects a device that has not entered a pairing code, 6-3 rejects a request without the signature of Cloudflare Access, and ComfyUI itself cannot be reached from outside. The tunnel is a [Cloudflare Tunnel](https://developers.cloudflare.com/tunnel/) (a [Quick Tunnel](https://developers.cloudflare.com/tunnel/get-started/quick-tunnels/) for 6-2).
- ComfyUI and custom nodes run as a dedicated user with reduced privileges that cannot reach the Drive mount or the connection secrets. This is not container isolation: the management services Colab itself runs are reachable by that user, so it does not stop a malicious node written to target Colab.
- Anyone who can connect can install custom nodes, and the nodes in the saved environment run again next time. If you installed a node you do not trust, turn off `RESTORE_SAVED_ENVIRONMENT` in cell 1 and start with a fresh environment.
- Do not use the "Mount Drive" button of Colab; connect Drive only through the setup in this notebook.
