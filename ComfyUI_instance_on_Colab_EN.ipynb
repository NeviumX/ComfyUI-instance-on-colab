# ComfyUI instance on Colab

Starts ComfyUI on a Colab GPU and lets you connect from the **Remote Connection** of [Comfy Desktop](https://github.com/Comfy-Org/Comfy-Desktop) or from a regular browser. Models and the environment are kept in `ComfyUI-Colab/` on Google Drive.

**How to use:** Choose a GPU in the runtime type, then run cells **1 → 2 → 3 → one of 6-1 to 6-3** in order. Set the fields and checkboxes of a cell before you run it. An action with a checkbox only happens when you run the cell with the box checked.

**Before you start:** Connecting Drive grants access to your whole Drive. Use a dedicated Google account that holds no personal data, and install only custom nodes and models you trust (see "Security" at the end). Follow the [Colab terms](https://research.google.com/colaboratory/faq.html); a paid plan is assumed.


```python
#@title 0. Check for updates
#@markdown Checks the latest release of ComfyUI. Turn this on to use it instead of `COMFY_REF` in cell 1. 
USE_LATEST_COMFYUI = False #@param {type:"boolean"}
import re, subprocess
found = subprocess.run(["git", "ls-remote", "--tags", "--refs", "https://github.com/Comfy-Org/ComfyUI.git", "v*"],
                       capture_output=True, text=True, timeout=60)
releases = [tuple(map(int, tag.groups())) for tag in
            (re.search(r"refs/tags/v(\d+)\.(\d+)\.(\d+)$", line) for line in found.stdout.splitlines()) if tag]
if found.returncode or not releases:
    raise RuntimeError("Could not get the latest release. Try again in a while.")
LATEST_COMFY_REF = "v" + ".".join(map(str, max(releases)))
print("Latest ComfyUI release:", LATEST_COMFY_REF)
# Cell 1 sets COMFY_REF from its form, so it takes the choice made here from this name.
COMFY_REF_OVERRIDE = LATEST_COMFY_REF if USE_LATEST_COMFYUI else ""
if USE_LATEST_COMFYUI:
    COMFY_REF = LATEST_COMFY_REF
    print("COMFY_REF is set to the latest release. Run cell 1 next.")
else:
    print("COMFY_REF keeps the value entered in cell 1. To use the latest release, check USE_LATEST_COMFYUI and run the cell.")
```

`ComfyUI-Colab/models/` on Google Drive has a folder for each type (`checkpoints`, `diffusion_models`, `text_encoders`, `vae`, `loras`, `controlnet` and so on).

- **Upload by hand:** upload into the matching folder on Google Drive, then run cell 5.
- **Download from a URL:** paste the URL into cell 4 and run it. The model is saved to Google Drive and can be used right away. With `RUNTIME_ONLY` checked it is not saved to Google Drive and is used on this runtime only.
- Models on Google Drive are fetched when they are used, so the first load of each one takes longer. To copy them all to Colab during setup instead, set `MODEL_ACCESS` to `copy` in cell 1.


```python
#@title 4. Download a model from a URL (Hugging Face / Civitai)
#@markdown Paste the URL of the file page on Hugging Face opened under Files and versions, or on Civitai the page with a version selected or the download link.
MODEL_URL = "" #@param {type:"string"}
#@markdown Destination folder (you can type one that is not in the list).
MODEL_FOLDER = "checkpoints" #@param ["checkpoints", "diffusion_models", "text_encoders", "vae", "loras", "controlnet", "clip_vision", "upscale_models", "embeddings"] {allow-input: true}
#@markdown File name (leave empty to keep the name from the site).
MODEL_FILENAME = "" #@param {type:"string"}
#@markdown SHA256 published by the site (optional; verified when given).
MODEL_SHA256 = "" #@param {type:"string"}
#@markdown Models that need authentication: register `HF_TOKEN` or `CIVITAI_TOKEN` in Colab Secrets (the key icon) and turn this on.
USE_DOWNLOAD_TOKEN = False #@param {type:"boolean"}
#@markdown Use it on this runtime only, without saving it to Google Drive. It is gone when the runtime is deleted.
RUNTIME_ONLY = False #@param {type:"boolean"}
DOWNLOAD_NOW = False #@param {type:"boolean"}
if DOWNLOAD_NOW:
    config = runtime.load_config()
    token = ""
    try:
        if USE_DOWNLOAD_TOKEN:
            from google.colab import userdata
            token = userdata.get(runtime.model_token_name(MODEL_URL))
        runtime.fetch_model(config, MODEL_URL, MODEL_FOLDER, MODEL_FILENAME, MODEL_SHA256, token,
                            runtime_only=RUNTIME_ONLY)
    finally:
        token = ""
else:
    print("To download, enter a URL, check DOWNLOAD_NOW and run the cell.")
```

```python
#@title 5. Load models you uploaded to Google Drive
REFRESH_MODELS = False #@param {type:"boolean"}
if REFRESH_MODELS:
    config = runtime.load_config()
    runtime.sync_models(config)
    print("If you are connected, reload the ComfyUI page. If you replaced a model that has the same name, stop with cell 10 (FLUSH_AND_UNMOUNT off) and run one of cells 6-1 to 6-3 again.")
else:
    print("After adding models to Google Drive by hand, check REFRESH_MODELS and run the cell.")
```

## Starting and connecting

Choose one way to connect and run its cell. To switch to another, stop with cell 10 (`FLUSH_AND_UNMOUNT` off) first. The first load can take about a minute.

- **6-1 Private URL:** for the browser this notebook is open in. Only a browser signed in to the Google account that runs this notebook can open it. Nothing is exposed to the outside. Comfy Desktop cannot connect this way.
- **6-2 Public URL:** for the **Remote Connection** of Comfy Desktop or another device. Open the URL that is shown; only a device where you enter the code shown in the notebook can connect. A code works once, for 10 minutes, and the URL stops working when you stop with cell 10.
- **6-3 Custom domain URL:** for Comfy Desktop or another device, at a fixed URL on your domain. Only accounts you allowed, signed in through Cloudflare Access, can connect. Needs a Cloudflare account, a domain and some setup beforehand.
- In Desktop, turn on **Auto-download outputs** in the connection settings to save generated files to your PC. If you used a browser, or some outputs were missed, collect them with cell 9.
- Once ComfyUI has started, its cell keeps running and waits, because if you only work on ComfyUI's page Colab may disconnect the notebook as idle. To use another cell, stop it with ■, then resume waiting with **6-4** when you are done.
- If the UI does not appear, the connection diagnosis in cell 7 shows where it stops.
- The **Stop** button at the top of the ComfyUI page also saves the environment and disconnects. You can choose to delete the runtime as well.


```python
#@title 6-1. Start and connect: Private URL (this browser only)
config = runtime.load_config()
runtime.start(config, "colab")
runtime.keep_running()
```

```python
#@title 6-2. Start and connect: Public URL
#@markdown After starting with 6-2, check this and run the cell to issue a new pairing code, for another device or when the code has expired.
NEW_PAIRING_CODE = False #@param {type:"boolean"}
config = runtime.load_config()
if NEW_PAIRING_CODE:
    runtime.issue_pairing_code(config)
else:
    runtime.start(config, "tunnel")
runtime.keep_running()
```

```python
#@title 6-3. Start and connect: Custom domain URL
#@markdown Beforehand, in Cloudflare Zero Trust, create a named tunnel (public hostname pointing to `http://127.0.0.1:8189`) and an Access application that protects that hostname. In Colab Secrets, register the tunnel token (the string that starts with `eyJ`) as `CF_TUNNEL_TOKEN` and the Application Audience (AUD) Tag of the Access application (64 characters) as `CF_ACCESS_AUD`, and allow this notebook to read them.
#@markdown
#@markdown Public hostname of the tunnel (for example `comfy.example.com`).
ACCESS_HOSTNAME = "" #@param {type:"string"}
config = runtime.load_config()
from google.colab import userdata
token, audience = userdata.get("CF_TUNNEL_TOKEN"), userdata.get("CF_ACCESS_AUD")
try:
    runtime.start_with_access(config, ACCESS_HOSTNAME, audience, token)
finally:
    token = audience = ""
runtime.keep_running()
```

```python
#@title 6-4. Resume waiting (prevents disconnection)
#@markdown Cells 6-1 to 6-3 keep running after ComfyUI has started, and wait while you use it. Colab cannot see what you do on ComfyUI's page, so if nothing happens in the notebook it may disconnect the runtime as idle; while a cell is running, the notebook is less likely to be taken for idle. To use another cell, stop the running cell (■) first, then run this cell to resume waiting when you are done. Stopping it does not stop ComfyUI. Do not close the Colab tab.
runtime.keep_running()
```

```python
#@title 7. Save status and logs / diagnosis
#@markdown When the UI does not appear, check this and run the cell to find where the connection stops.
RUN_CONNECTION_DIAGNOSIS = False #@param {type:"boolean"}
if RUN_CONNECTION_DIAGNOSIS:
    runtime.run_diagnosis()
path = runtime.paths.PRIVATE / "checkpoint-status.json"
print(path.read_text() if path.exists() else "No save status yet.")
for name in ("comfy.log", "supervisor.log", "proxy.log", "tunnel.log"):
    path = runtime.paths.PRIVATE / name
    if path.exists():
        print("---", name)
        print(runtime.redact(path.read_text(errors="replace")[-5000:]))
```

## Saving and finishing

ComfyUI itself, custom nodes and settings is saved to Google Drive automatically and restored by cell 3 next time. Use cell 8 when you want to be sure the latest changes are saved. Generated files are not saved to Google Drive.

When you finish, make sure your outputs are on your PC, run cell 10, and wait for its completion message before you delete the runtime.


```python
#@title 8. Save the environment to Google Drive now
SAVE_NOW = False #@param {type:"boolean"}
if SAVE_NOW:
    config = runtime.load_config()
    runtime.checkpoint(config, force=True)
    print("Saved the environment and new models to Google Drive.")
else:
    print("To save, check SAVE_NOW and run the cell.")
```

```python
#@title 9. Download generated files to your PC as a ZIP
DOWNLOAD_OUTPUT_ZIP = False #@param {type:"boolean"}
if DOWNLOAD_OUTPUT_ZIP:
    from google.colab import files
    files.download(runtime.zip_outputs())
else:
    print("To collect them, check DOWNLOAD_OUTPUT_ZIP and run the cell.")
```

```python
#@title 10. Save and stop
STOP_NOW = False #@param {type:"boolean"}
#@markdown Finish writing to Google Drive and disconnect it (on before deleting the runtime; off if you will start again from one of cells 6-1 to 6-3 on the same runtime).
FLUSH_AND_UNMOUNT = True #@param {type:"boolean"}
if STOP_NOW:
    config = runtime.load_config()
    runtime.stop(config, save=True)
    if FLUSH_AND_UNMOUNT:
        from google.colab import drive
        runtime.hide_drive_models()
        drive.flush_and_unmount()
        print("Finished writing to Google Drive. You can delete the runtime.")
    else:
        print("Google Drive is still connected. Before deleting the runtime, run this cell with FLUSH_AND_UNMOUNT checked.")
else:
    print("To finish, check STOP_NOW and run the cell.")
```

## Security

- Cell 6-1 exposes nothing to the outside. With cells 6-2 and 6-3 only an authenticated entrance is exposed: 6-2 rejects a device that has not entered a pairing code, and 6-3 a device that has not signed in through Cloudflare Access.
- Custom nodes run their own code on the runtime. The notebook keeps Google Drive and the connection secrets away from them, but this is not complete isolation. Install only custom nodes and models you trust.
- Anyone who can connect can install custom nodes, and the nodes in the saved environment run again next time. If you installed a node you do not trust, turn off `RESTORE_SAVED_ENVIRONMENT` in cell 1 and start with a fresh environment.
- Do not use the "Mount Drive" button of Colab; connect Drive only through the setup in this notebook.
